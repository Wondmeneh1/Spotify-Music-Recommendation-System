## Scale the features

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

feature_matrix = scaler.fit_transform(df[audio_features])
```

```python
print(feature_matrix.shape)
```

```python
from sklearn.neighbors import NearestNeighbors

model = NearestNeighbors(
    n_neighbors=11,
    metric='cosine',
    algorithm='brute'
)

model.fit(feature_matrix)
```

### . Building and Testing the Song Recommender

Now, let's write a function `recommend_songs` that utilizes our trained model to find and display the top 10 most similar tracks for any given song in the dataset.

```python
def recommend_songs(song_name, df, model, feature_matrix, n_recommendations=10):
    # Find matches for the song name (case-insensitive search)
    matches = df[df['track_name'].str.lower() == song_name.lower()]
    
    if matches.empty:
        print(f"Song '{song_name}' not found in the dataset. Here are some options containing your search term:")
        partial_matches = df[df['track_name'].str.contains(song_name, case=False, na=False)]
        if not partial_matches.empty:
            display(partial_matches[['track_name', 'track_artist']].head(10))
        else:
            print("No close matches found.")
        return None
    
    # Get the index of the first match
    query_index = matches.index[0]
    query_song_details = df.iloc[query_index]
    
    # Reshape the feature vector for prediction
    query_vector = feature_matrix[query_index].reshape(1, -1)
    
    # Find the nearest neighbors (we ask for n+1 since the song itself will be the closest match)
    distances, indices = model.kneighbors(query_vector, n_neighbors=n_recommendations + 1)
    
    # Extract recommendation indices and distances (excluding the first one which is the query song itself)
    rec_indices = indices.flatten()[1:]
    rec_distances = distances.flatten()[1:]
    
    # Build a DataFrame of recommendations
    recommendations = df.iloc[rec_indices].copy()
    recommendations['similarity_distance'] = rec_distances
    
    print(f"Recommendations for '{query_song_details['track_name']}' by {query_song_details['track_artist']}:\n")
    
    cols_to_show = ['track_name', 'track_artist', 'playlist_genre', 'playlist_subgenre', 'similarity_distance']
    display(recommendations[cols_to_show])
    
    return recommendations
```

# Let's test the recommender system with a song from the dataset

```python

test_song = "Someone You Loved - Future Humans Remix"
recommendations = recommend_songs(test_song, df, model, feature_matrix)
```

###  Refining the Recommender (Removing Duplicate Songs)

Let's filter out duplicate tracks based on `track_name` and `track_artist` to ensure that every recommendation is unique.

```python
# Drop duplicates based on track name and artist
df_unique = df.drop_duplicates(subset=['track_name', 'track_artist']).reset_index(drop=True)

# Scale features for the unique songs dataset
feature_matrix_unique = scaler.fit_transform(df_unique[audio_features])

# Retrain the Nearest Neighbors model
model_unique = NearestNeighbors(n_neighbors=11, metric='cosine', algorithm='brute')
model_unique.fit(feature_matrix_unique)

# Test the recommender with the same song on the deduplicated dataset
recommendations_unique = recommend_songs(test_song, df_unique, model_unique, feature_matrix_unique)
```
